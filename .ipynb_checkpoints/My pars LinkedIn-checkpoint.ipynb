```python
import requests
from bs4 import BeautifulSoup
import time
import random
import json
from urllib.parse import quote
import datetime

# --- Configurare ---
headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'}
jobs_list = []
unique_job_links = set()
output_file = 'E-Commerce_jobs_germany.json'
autosave_interval = 300 

germany_cities = [
    "Germany"
]

keywords = [
    "E-Commerce Manager", "E-Commerce Specialist", "Online Marketing Manager",
    "Online Sales Manager", "Marketplace Manager", "Amazon Manager",
    "Shop Manager", "E-Commerce Spezialist", "Online-Marketing Manager",
    "Online Vertrieb", "Marketplace Spezialist", "Amazon Spezialist"
]

def save_data_to_json(data, filename):
    """Salvează lista de dicționare într-un fișier JSON, suprascriindu-l."""
    with open(filename, 'w', encoding='utf-8') as f:
        json.dump(data, f, ensure_ascii=False, indent=4)

def get_user_choice():
    """Afișează meniul și returnează o listă de filtre de procesat."""
    print("🤖 Bun venit! Alegeți perioada pentru căutarea de joburi:")
    print("="*50)
    print("1: Ultima lună (o singură căutare)")
    print("2: Ultimele 3 luni (rulează 3 căutări: pt 1, 2, și 3 luni)")
    print("3: Ultimele 6 luni (rulează 6 căutări: pt 1, 2...6 luni)")
    print("4: Perioadă personalizată (ex: ultimele 4 luni)")
    print("5: Oricând (toate joburile, o singură căutare)")
    print("="*50)

    time_options = {
        "1_month": {"label": "Ultima lună", "seconds": "r2592000"},
        "anytime": {"label": "Oricând", "seconds": ""}
    }

    filters_to_process = []

    while True:
        choice = input("Introduceți numărul opțiunii (1-5): ")
        if choice == "1":
            print("\n✅ Ați ales: Căutare pentru ultima lună.")
            filters_to_process.append(time_options["1_month"])
            break
        elif choice == "2":
            print("\n✅ Ați ales: Căutare secvențială pentru ultimele 3 luni.")
            for i in range(1, 4):
                seconds = i * 2592000
                filters_to_process.append({"label": f"Ultimele {i} luni", "seconds": f"r{seconds}"})
            break
        elif choice == "3":
            print("\n✅ Ați ales: Căutare secvențială pentru ultimele 6 luni.")
            for i in range(1, 7):
                seconds = i * 2592000
                filters_to_process.append({"label": f"Ultimele {i} luni", "seconds": f"r{seconds}"})
            break
        elif choice == "4":
            while True:
                try:
                    num_months = int(input("Introduceți numărul de luni pentru căutare (ex: 4): "))
                    if num_months > 0:
                        print(f"\n✅ Ați ales: Căutare secvențială pentru ultimele {num_months} luni.")
                        for i in range(1, num_months + 1):
                            seconds = i * 2592000
                            filters_to_process.append({"label": f"Ultimele {i} luni", "seconds": f"r{seconds}"})
                        break
                    else:
                        print("❌ Vă rugăm introduceți un număr pozitiv.")
                except ValueError:
                    print("❌ Input invalid. Vă rugăm introduceți un număr întreg.")
            break
        elif choice == "5":
            print("\n✅ Ați ales: Căutare pentru 'Oricând'.")
            filters_to_process.append(time_options["anytime"])
            break
        else:
            print("❌ Opțiune invalidă. Vă rugăm alegeți un număr între 1 și 5.")
            
    return filters_to_process

# --- Etapa 1: Colectarea listei de joburi ---
filters_to_run = get_user_choice()

time_start = time.time()
last_save_time = time_start

for filter_details in filters_to_run:
    time_param = filter_details['seconds']
    time_label = filter_details['label']
    
    print(f"\n--- 🔍 Încep căutarea pentru perioada: '{time_label}' ---")

    for city in germany_cities:
        for kw in keywords:
            for start in range(0, 250, 25): 
                kw_encoded = quote(kw)
                city_encoded = quote(city)
                url = f'https://www.linkedin.com/jobs/search/?keywords={kw_encoded}&location={city_encoded}&f_TPR={time_param}&start={start}'

                try:
                    response = requests.get(url, headers=headers, timeout=15)
                    response.raise_for_status()
                    
                    soup = BeautifulSoup(response.text, 'html.parser')
                    job_cards = soup.find_all('div', class_='base-card')
                    
                    if not job_cards:
                        break

                    new_jobs_found_on_page = 0
                    for job in job_cards:
                        link_element = job.find('a', class_='base-card__full-link')
                        
                        # ### MODIFICAREA CHEIE AICI ### 
                        # Curățăm link-ul de parametrii de tracking pentru a evita duplicatele
                        link = link_element['href'].split('?')[0] if link_element else ''

                        if link and link not in unique_job_links:
                            unique_job_links.add(link) 
                            new_jobs_found_on_page += 1

                            title_element = job.find('h3', class_='base-search-card__title')
                            company_element = job.find('h4', class_='base-search-card__subtitle')
                            location_element = job.find('span', class_='job-search-card__location')

                            title = title_element.text.strip() if title_element else ''
                            company = company_element.text.strip() if company_element else ''
                            location = location_element.text.strip() if location_element else ''
                            
                            jobs_list.append({
                                'Title': title, 'Company': company, 'Location': location,
                                'Link': link, 'Keyword': kw, 'City_Searched': city,
                                'Time_Filter': time_label
                            })
                    
                    if new_jobs_found_on_page > 0:
                        print(f"✅ Găsite {new_jobs_found_on_page} joburi noi pentru '{kw}' (Pagina {start//25 + 1})")
                    
                    time.sleep(random.uniform(3, 7))
                    
                    if time.time() - last_save_time > autosave_interval:
                        print(f"\n--- ⏱️ AUTOSAVE: Salvez {len(jobs_list)} joburi în '{output_file}'... ---")
                        save_data_to_json(jobs_list, output_file)
                        last_save_time = time.time()

                except requests.exceptions.RequestException as e:
                    print(f"❌ Eroare la accesarea URL-ului {url}: {e}")
                    time.sleep(15)

# --- Etapa 2: Extragerea descrierilor ---
if jobs_list:
    print(f"\n📝 Extrag descrierile detaliate pentru {len(jobs_list)} joburi unice găsite...")
    total_jobs = len(jobs_list)

    for i, job in enumerate(jobs_list):
        if 'Description' in job: 
            continue

        if not job['Link'] or not job['Link'].startswith('http'):
            job['Description'] = 'Link invalid'
            continue

        url = job['Link']
        try:
            response = requests.get(url, headers=headers, timeout=12)
            response.raise_for_status()
            
            soup = BeautifulSoup(response.text, 'html.parser')
            desc_block = soup.find('div', class_='description__text')
            
            description = desc_block.get_text(separator=' ', strip=True) if desc_block else 'Descriere nu a fost gasita'
                
        except requests.exceptions.RequestException as e:
            print(f"❌ Eroare la extragerea descrierii de la {url}: {e}")
            description = 'Eroare la extragere'

        job['Description'] = description
        print(f"    -> Procesat {i + 1}/{total_jobs}: {job['Title'][:40]}...")
        
        time.sleep(random.uniform(2, 5))
        
        if time.time() - last_save_time > autosave_interval:
            print(f"\n--- ⏱️ AUTOSAVE: Salvez {len(jobs_list)} joburi în '{output_file}'... ---")
            save_data_to_json(jobs_list, output_file)
            last_save_time = time.time()
else:
    print("\n🤷‍♂️ Nu au fost găsite joburi noi. Scriptul se va încheia.")

# --- Etapa 3: Salvarea finală în fișierul JSON ---
if jobs_list:
    print(f"\n💾 Salvare finală: Salvez toate cele {len(jobs_list)} rezultate în fișierul '{output_file}'...")
    save_data_to_json(jobs_list, output_file)
    print(f"\n✅ Gata! Datele au fost salvate în fișierul '{output_file}'.")

time_end = time.time()
total_duration_seconds = time_end - time_start
duration_formatted = str(datetime.timedelta(seconds=total_duration_seconds))

print(f"\n⏱️ Cautarea a durat: {duration_formatted}")
```

```python
import requests
from bs4 import BeautifulSoup
import time
import random
import json
from urllib.parse import quote
import datetime

# --- Configurare ---
headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'}
jobs_list = []
unique_job_links = set()
output_file = 'E-Commerce_jobs_germany.json'
autosave_interval = 300  # 5 minute

germany_cities = ["Germany"]
keywords = [
    "E-Commerce Manager", "E-Commerce Specialist", "Online Marketing Manager",
    "Online Sales Manager", "Marketplace Manager", "Amazon Manager",
    "Shop Manager", "E-Commerce Spezialist", "Online-Marketing Manager",
    "Online Vertrieb", "Marketplace Spezialist", "Amazon Spezialist"
]

def save_data_to_json(data, filename):
    with open(filename, 'w', encoding='utf-8') as f:
        json.dump(data, f, ensure_ascii=False, indent=4)

def get_user_choice():
    # ... (Funcția rămâne neschimbată, nu o mai includ aici pentru a economisi spațiu)
    print("🤖 Bun venit! Alegeți perioada pentru căutarea de joburi:")
    print("="*50)
    print("1: Ultima lună (o singură căutare)")
    print("2: Ultimele 3 luni (rulează 3 căutări: pt 1, 2, și 3 luni)")
    print("3: Ultimele 6 luni (rulează 6 căutări: pt 1, 2...6 luni)")
    print("4: Perioadă personalizată (ex: ultimele 4 luni)")
    print("5: Oricând (toate joburile, o singură căutare)")
    print("="*50)
    time_options = {
        "1_month": {"label": "Ultima lună", "seconds": "r2592000"},
        "anytime": {"label": "Oricând", "seconds": ""}
    }
    filters_to_process = []
    while True:
        choice = input("Introduceți numărul opțiunii (1-5): ")
        if choice == "1":
            print("\n✅ Ați ales: Căutare pentru ultima lună.")
            filters_to_process.append(time_options["1_month"])
            break
        elif choice == "2":
            print("\n✅ Ați ales: Căutare secvențială pentru ultimele 3 luni.")
            for i in range(1, 4):
                seconds = i * 2592000
                filters_to_process.append({"label": f"Ultimele {i} luni", "seconds": f"r{seconds}"})
            break
        elif choice == "3":
            print("\n✅ Ați ales: Căutare secvențială pentru ultimele 6 luni.")
            for i in range(1, 7):
                seconds = i * 2592000
                filters_to_process.append({"label": f"Ultimele {i} luni", "seconds": f"r{seconds}"})
            break
        elif choice == "4":
            while True:
                try:
                    num_months = int(input("Introduceți numărul de luni pentru căutare (ex: 4): "))
                    if num_months > 0:
                        print(f"\n✅ Ați ales: Căutare secvențială pentru ultimele {num_months} luni.")
                        for i in range(1, num_months + 1):
                            seconds = i * 2592000
                            filters_to_process.append({"label": f"Ultimele {i} luni", "seconds": f"r{seconds}"})
                        break
                    else:
                        print("❌ Vă rugăm introduceți un număr pozitiv.")
                except ValueError:
                    print("❌ Input invalid. Vă rugăm introduceți un număr întreg.")
            break
        elif choice == "5":
            print("\n✅ Ați ales: Căutare pentru 'Oricând'.")
            filters_to_process.append(time_options["anytime"])
            break
        else:
            print("❌ Opțiune invalidă. Vă rugăm alegeți un număr între 1 și 5.")
    return filters_to_process

# ### MODIFICARE ### Funcție pentru a extrage toate detaliile de pe pagina unui job
def extract_job_details(job_soup):
    """Extrage detalii specifice de pe pagina individuală a unui job."""
    details = {
        'description': 'N/A',
        'publishedAt': 'N/A',
        'contractType': 'N/A',
        'workType': 'N/A',
        'sector': 'N/A'
    }

    # 1. Extrage Descrierea
    desc_block = job_soup.find('div', class_='description__text')
    if desc_block:
        details['description'] = desc_block.get_text(separator='\n', strip=True)

    # 2. Extrage Data postării
    posted_time_element = job_soup.find('span', class_='posted-time-ago__text')
    if posted_time_element:
        details['publishedAt'] = posted_time_element.text.strip()
        
    # 3. Extrage Criteriile (Tip contract, Tip muncă, Sector etc.)
    criteria_list = job_soup.find('ul', class_='description__job-criteria-list')
    if criteria_list:
        criteria_items = criteria_list.find_all('li', class_='description__job-criteria-item')
        for item in criteria_items:
            header = item.find('h3', class_='description__job-criteria-subheader')
            text = item.find('span', class_='description__job-criteria-text')
            if header and text:
                header_text = header.text.strip().lower()
                text_value = text.text.strip()
                
                if 'employment type' in header_text:
                    details['contractType'] = text_value
                elif 'job function' in header_text:
                    details['workType'] = text_value
                elif 'industries' in header_text:
                    details['sector'] = text_value
    
    return details


# --- Etapa Unică: Căutare, Extragere și Salvare ---
filters_to_run = get_user_choice()
time_start = time.time()
last_save_time = time_start
total_jobs_processed = 0

for filter_details in filters_to_run:
    time_param = filter_details['seconds']
    time_label = filter_details['label']
    print(f"\n--- 🔍 Încep căutarea pentru perioada: '{time_label}' ---")

    for city in germany_cities:
        for kw in keywords:
            for start in range(0, 250, 25): 
                search_url = f'https://www.linkedin.com/jobs/search/?keywords={quote(kw)}&location={quote(city)}&f_TPR={time_param}&start={start}'

                try:
                    response = requests.get(search_url, headers=headers, timeout=15)
                    response.raise_for_status()
                    soup = BeautifulSoup(response.text, 'html.parser')
                    job_cards = soup.find_all('div', class_='base-card')
                    
                    if not job_cards:
                        break

                    for job in job_cards:
                        link_element = job.find('a', class_='base-card__full-link')
                        clean_link = link_element['href'].split('?')[0] if link_element else ''

                        if clean_link and clean_link not in unique_job_links:
                            unique_job_links.add(clean_link)
                            total_jobs_processed += 1
                            
                            print(f"    -> Procesez jobul {total_jobs_processed}: {clean_link}")

                            # ### MODIFICARE ### Vizităm direct linkul jobului
                            try:
                                job_response = requests.get(clean_link, headers=headers, timeout=12)
                                job_response.raise_for_status()
                                job_soup = BeautifulSoup(job_response.text, 'html.parser')
                                
                                # Extragem detaliile de bază de pe card
                                title = job.find('h3', class_='base-search-card__title').text.strip()
                                company = job.find('h4', class_='base-search-card__subtitle').text.strip()
                                location = job.find('span', class_='job-search-card__location').text.strip()
                                
                                # Extragem detaliile complete de pe pagina jobului
                                job_details = extract_job_details(job_soup)

                                jobs_list.append({
                                    'Title': title,
                                    'Company': company,
                                    'Location': location,
                                    'Link': clean_link,
                                    'Keyword': kw,
                                    'City_Searched': city,
                                    'Time_Filter': time_label,
                                    'Published_At': job_details['publishedAt'],
                                    'Contract_Type': job_details['contractType'],
                                    'Work_Type': job_details['workType'],
                                    'Sector': job_details['sector'],
                                    'Description': job_details['description']
                                })
                                
                                time.sleep(random.uniform(8, 18)) # Pauză după fiecare extragere de detalii

                            except requests.exceptions.RequestException as e:
                                print(f"      ❌ Eroare la accesarea paginii jobului {clean_link}: {e}")
                                # Adăugăm jobul cu date parțiale în caz de eroare la extragere
                                jobs_list.append({
                                    'Title': title, 'Company': company, 'Location': location,
                                    'Link': clean_link, 'Keyword': kw, 'City_Searched': city,
                                    'Time_Filter': time_label, 'Description': 'Eroare la extragere detalii'
                                })
                    
                    if time.time() - last_save_time > autosave_interval:
                        print(f"\n--- ⏱️ AUTOSAVE: Salvez {len(jobs_list)} joburi în '{output_file}'... ---")
                        save_data_to_json(jobs_list, output_file)
                        last_save_time = time.time()

                except requests.exceptions.RequestException as e:
                    print(f"❌ Eroare la accesarea paginii de căutare {search_url}: {e}")
                    time.sleep(15)

# --- Salvarea finală ---
if jobs_list:
    print(f"\n💾 Salvare finală: Salvez toate cele {len(jobs_list)} rezultate în fișierul '{output_file}'...")
    save_data_to_json(jobs_list, output_file)
    print(f"\n✅ Gata! Datele au fost salvate în fișierul '{output_file}'.")
else:
    print("\n🤷‍♂️ Nu au fost găsite joburi. Scriptul se va încheia.")

time_end = time.time()
total_duration_seconds = time_end - time_start
duration_formatted = str(datetime.timedelta(seconds=total_duration_seconds))
print(f"\n⏱️ Cautarea a durat: {duration_formatted}")
```
